# Europe: employment rates by citizenship

Employment rates of nationals, EU citizens and non-EU citizens (age 20–64): the labour-market integration gap.

**Datasets:** `eurostat/lfsa_ergan`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["eurostat/lfsa_ergan"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
g = load("eurostat/lfsa_ergan")
print({c: g[c].drop_duplicates().head(12).tolist() for c in ("citizen", "age", "sex", "unit") if c in g.columns})
```

```python
x = select(g, sex="T", age="Y20-64", citizen=["NAT", "EU27_2020_FOR", "NEU27_2020_FOR"])
areas = ["EU27_2020", "DE", "FR", "ES", "IT", "SE", "NL"]
last = x[x.ref_area.isin(areas)].sort_values("period").groupby(["ref_area", "citizen"]).tail(1)
last.pivot_table(index="ref_area", columns="citizen", values="value").assign(gap_non_eu=lambda d: d["NEU27_2020_FOR"] - d["NAT"])
```
