# Remittances

Personal remittances received: the largest recipients and long-run growth. Remittances are the income channel through which labour migration reaches origin economies.

**Datasets:** `worldbank/remittances_received`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["worldbank/remittances_received"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
r = load("worldbank/remittances_received")
r = r[r.ref_area.str.len() == 3]
last = r.dropna(subset=["value"]).sort_values("period").groupby("ref_area").tail(1)
top = last.nlargest(15, "value").set_index("area_name")
(top["value"] / 1e9).round(1).to_frame("US$ bn")
```

```python
p = pivot(select(r, ref_area=["IND", "MEX", "PHL", "CHN", "EGY", "PAK", "BGD", "NGA"]))
plot_lines(p.loc["1990":] / 1e9, "Remittances received", "US$ bn", "World Bank (BX.TRF.PWKR.CD.DT)", names={});
```
