# International migrant stock

Migrant stock by destination over time (UN DESA via the World Bank), and the full origin–destination matrix from the UN DESA 2024 release.

**Datasets:** `worldbank/migrant_stock`, `undesa/migrant_stock_2024`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["worldbank/migrant_stock", "undesa/migrant_stock_2024"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
m = load("worldbank/migrant_stock")
latest(select(m, ref_area=FOCUS))
```

```python
p = pivot(select(m, ref_area=[a for a in FOCUS if a != "WLD"]))
plot_lines(p / 1e6, "International migrant stock", "millions", "UN DESA via World Bank (SM.POP.TOTL)");
```
