# OECD average annual wages

Average annual wages per full-time-equivalent employee, constant prices and PPP; real growth since 2000.

**Datasets:** `oecd/average_annual_wages`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["oecd/average_annual_wages"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
w = load("oecd/average_annual_wages")
print({c: w[c].drop_duplicates().head(10).tolist() for c in w.columns if c not in ("value", "period", "ref_area")})
```

```python
cols = [c for c in ("unit_measure", "price_base", "unit_mult") if c in w.columns]
print("Filter on these to pick constant-price PPP USD:", cols)
sel = {}
if "price_base" in w.columns: sel["price_base"] = "Q"
real = select(w, **sel) if sel else w
p = pivot(select(real, ref_area=[a for a in FOCUS if a in set(real.ref_area)]))
plot_lines(p / p.loc[p.index.astype(str) == "2000"].iloc[0] * 100, "Real average wages, 2000 = 100", "index", "OECD Average annual wages");
```
