# Earnings by economic activity

Average monthly earnings of employees by activity (ILO national series): how pay in health, education, construction, utilities and information compares with the economy-wide mean in each country.

**Datasets:** `ilostat/mean_monthly_earnings`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["ilostat/mean_monthly_earnings"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
e = load("ilostat/mean_monthly_earnings")
print(e["classif1"].drop_duplicates().tolist()[:40])
```

```python
e = select(e, sex="SEX_T")
acts = [c for c in e.classif1.unique() if c.startswith("ECO_ISIC4_")]
rel = e[e.classif1.isin(acts + ["ECO_ISIC4_TOTAL"])]
last = rel.sort_values("period").groupby(["ref_area", "classif1"]).tail(1)
tab = last.pivot_table(index="ref_area", columns="classif1", values="value")
(tab.div(tab["ECO_ISIC4_TOTAL"], axis=0)).loc[[a for a in FOCUS if a in tab.index]].round(2)
```
