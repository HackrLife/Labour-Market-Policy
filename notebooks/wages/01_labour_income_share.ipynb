# Labour income share

Labour's share of GDP (ILO modelled estimates, including an imputation for the self-employed) and its change since 2004.

**Datasets:** `ilostat/labour_income_share`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["ilostat/labour_income_share"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
ls = select(load("ilostat/labour_income_share"), ref_area=FOCUS)
plot_lines(pivot(ls), "Labour income share, % of GDP", "%", "ILO modelled estimates (LAP_2GDP_NOC_RT_A)");
```

```python
p = pivot(ls)
(p.iloc[-1] - p.loc[p.index.astype(str) == "2004"].iloc[0]).sort_values().to_frame(f"change 2004–{p.index[-1]} (pp)")
```
