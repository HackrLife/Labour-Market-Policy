# House prices relative to income

OECD price-to-income ratios (index) and the curated Demographia median multiples for major cities.

**Datasets:** `oecd/house_prices`  
**Curated inputs:** `housing/housing_affordability_policy`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["oecd/house_prices"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
h = load("oecd/house_prices")
print({c: h[c].drop_duplicates().head(12).tolist() for c in h.columns if c not in ("value", "period", "ref_area", "source", "dataset")})
```

```python
m = [c for c in h.columns if c in ("measure", "indicator")]
sel = {m[0]: "HPI_YDH"} if m and "HPI_YDH" in set(h[m[0]]) else {}
x = select(h, ref_area=[a for a in FOCUS if a in set(h.ref_area)], **sel)
plot_lines(pivot(to_annual(x)), "House price-to-income ratio (index)", "index", "OECD Analytical house price indicators");
```
