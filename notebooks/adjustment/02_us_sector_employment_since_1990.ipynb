# United States: sector employment since 1990

Levels and net change of payroll employment in health care, construction, manufacturing and information.

**Datasets:** `bls/headline_monthly`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["bls/headline_monthly"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
h = load("bls/headline_monthly")
sec = ["CES6562000001", "CES2000000001", "CES3000000001", "CES5000000001"]
s = pivot(select(h, indicator=sec), columns="indicator_label")
plot_lines(s / 1000, "Payroll employment", "millions", "BLS CES", names={});
```

```python
jan95 = s.loc["1995-01"]
last = s.iloc[-1]
pd.DataFrame({"Jan 1995": jan95, s.index[-1]: last, "net change": last - jan95, "% change": (last / jan95 - 1) * 100}).round(1)
```
