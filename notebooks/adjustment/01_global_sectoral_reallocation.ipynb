# Sectoral reallocation: agriculture, industry, services

Employment shares by broad sector since 1991 (ILO modelled estimates), and the long-run US series from 1900 for historical context.

**Datasets:** `worldbank/emp_agriculture_share`, `worldbank/emp_industry_share`, `worldbank/emp_services_share`  
**Curated inputs:** `adjustment/long_run_sectoral_employment`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["worldbank/emp_agriculture_share", "worldbank/emp_industry_share", "worldbank/emp_services_share"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
sec = {"Agriculture": "worldbank/emp_agriculture_share", "Industry": "worldbank/emp_industry_share", "Services": "worldbank/emp_services_share"}
frames = {k: pivot(select(load(v), ref_area=FOCUS)) for k, v in sec.items()}
fig, axes = plt.subplots(1, 3, figsize=(16, 5), sharey=True)
for ax, (k, p) in zip(axes, frames.items()):
    plot_lines(p[["WLD", "IND", "CHN", "IDN", "USA", "AUS"]], f"{k}, % of employment", "%", "ILO modelled via World Bank", ax=ax)
```

```python
chg = pd.DataFrame({k: p.iloc[-1] - p.iloc[0] for k, p in frames.items()})
chg.rename(index=FOCUS_NAMES).round(1)
```
