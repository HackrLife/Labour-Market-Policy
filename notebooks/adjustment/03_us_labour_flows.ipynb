# United States: hires, quits and layoffs

Gross labour flows from JOLTS. Annual hires of about 60 million set the scale against which any displacement estimate should be read; the quits rate tracks workers' confidence in moving between jobs.

**Datasets:** `bls/jolts`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["bls/jolts"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
j = load("bls/jolts")
f = pivot(select(j, indicator=["JTS000000000000000HIL", "JTS000000000000000QUL", "JTS000000000000000LDL"]), columns="indicator_label")
plot_lines(f / 1000, "Monthly hires, quits and layoffs", "millions", "BLS JOLTS", names={});
```

```python
annual = to_annual(select(j, indicator=["JTS000000000000000HIL", "JTS000000000000000QUL", "JTS000000000000000LDL"]))
annual["value"] = annual["value"] * 12 / 1000
annual.pivot_table(index="period", columns="indicator_label", values="value").tail(10).round(1)
```
