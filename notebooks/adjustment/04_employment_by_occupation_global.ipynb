How the occupational mix differs across economies: professionals, technicians and clerical workers (the information-intensive groups) against craft, plant and elementary occupations.

```python
o = select(load("ilostat/employment_by_occupation_modelled"), sex="SEX_T")
print(o.classif1.drop_duplicates().tolist())
```

```python
skill = o[o.classif1.str.startswith("OCU_ISCO08_") & ~o.classif1.str.endswith("TOTAL")]
last = skill[skill.period == skill.period.max()]
shares = last.pivot_table(index="ref_area", columns="classif1", values="value")
shares = shares.div(shares.sum(axis=1), axis=0) * 100
shares.loc[[a for a in FOCUS if a in shares.index]].round(1)
```
