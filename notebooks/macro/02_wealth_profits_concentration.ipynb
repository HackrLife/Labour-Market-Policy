# United States: wealth concentration and corporate profits

Distributional Financial Accounts (top 1% and bottom 50% shares of net worth) and after-tax corporate profits relative to GDP.

**Datasets:** `fred/distribution_and_profits`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["fred/distribution_and_profits"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
d = load("fred/distribution_and_profits")
w = pivot(select(d, indicator=["WFRBST01134", "WFRBSB50215"]), columns="indicator").rename(columns={"WFRBST01134": "Top 1%", "WFRBSB50215": "Bottom 50%"})
plot_lines(w, "Share of US household net worth", "%", "Federal Reserve DFA via FRED", names={});
```

```python
g = pivot(select(d, indicator=["CP", "GDP"]), columns="indicator")
prof = (g["CP"] / g["GDP"] * 100).dropna().to_frame("after-tax profits, % of GDP")
plot_lines(prof, "Corporate profits after tax, % of GDP", "%", "BEA via FRED (CP, GDP)", names={});
```
