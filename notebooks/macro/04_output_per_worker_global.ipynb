# Output per worker

Labour productivity levels (GDP per worker, PPP) and growth since 2000 across economies.

**Datasets:** `ilostat/output_per_worker`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["ilostat/output_per_worker"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
o = select(load("ilostat/output_per_worker"), ref_area=FOCUS)
p = pivot(o)
plot_bars(p.iloc[-1] / 1000, f"Output per worker, {p.index[-1]}", "thousand 2021 PPP $", "ILO modelled estimates");
```

```python
growth = ((p.iloc[-1] / p.loc[p.index.astype(str) == "2000"].iloc[0]) ** (1 / (int(str(p.index[-1])) - 2000)) - 1) * 100
growth.sort_values().rename(index=FOCUS_NAMES).to_frame("annual growth since 2000, %").round(2)
```
