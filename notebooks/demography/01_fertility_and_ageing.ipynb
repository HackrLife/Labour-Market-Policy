# Fertility and ageing

Total fertility rates against the 2.1 replacement level, and old-age dependency ratios: the labour-supply backdrop for every other notebook.

**Datasets:** `worldbank/fertility_rate`, `worldbank/age_dependency_old`  
**Curated inputs:** `demography/fertility_ageing_urbanisation`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["worldbank/fertility_rate", "worldbank/age_dependency_old"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
f = pivot(select(load("worldbank/fertility_rate"), ref_area=FOCUS))
ax = plot_lines(f.loc["1970":], "Total fertility rate", "births per woman", "UN WPP via World Bank (SP.DYN.TFRT.IN)")
ax.axhline(2.1, color="k", ls="--", lw=0.8);
```

```python
d = pivot(select(load("worldbank/age_dependency_old"), ref_area=FOCUS))
plot_bars(d.iloc[-1], f"Old-age dependency ratio, {d.index[-1]}", "65+ per 100 aged 15–64", "World Bank (SP.POP.DPND.OL)");
```
