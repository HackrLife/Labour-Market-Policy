# World Population Prospects 2024: projections

Medium-variant projections of population, fertility and working-age population to 2100 from the UN DESA release file.

**Datasets:** `undesa/wpp_2024_indicators`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["undesa/wpp_2024_indicators"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
w = pd.read_csv(raw_payload("undesa/wpp_2024_indicators"), low_memory=False)
print(w.columns.tolist()[:40])
```

```python
iso = ["WLD"] if "ISO3_code" not in w.columns else None
sel = w[w["ISO3_code"].isin(["AUS", "USA", "JPN", "KOR", "CHN", "IND", "IDN", "DEU", "ITA"])] if "ISO3_code" in w.columns else w
p = sel.pivot_table(index="Time", columns="ISO3_code", values="TPopulation1July")
plot_lines(p / 1000, "Population (medium variant)", "millions", "UN DESA WPP 2024", names={});
```
