```python
import zipfile
z = zipfile.ZipFile(raw_payload("research/onet_database"))
[n for n in z.namelist() if n.endswith(".txt")][:40]
```

```python
wc = pd.read_csv(z.open(next(n for n in z.namelist() if n.endswith("Work Context.txt"))), sep="\t")
phys = wc[wc["Element Name"].isin(["Physical Proximity", "Spend Time Using Your Hands to Handle, Control, or Feel Objects, Tools, or Controls", "Face-to-Face Discussions"]) & (wc["Scale ID"] == "CX")]
phys.pivot_table(index="O*NET-SOC Code", columns="Element Name", values="Data Value").describe()
```
