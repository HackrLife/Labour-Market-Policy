# Occupations: wages and projected growth, physical economy vs digital

BLS OEWS May 2025 wages and Employment Projections 2025–35 for physical-economy white-collar occupations, blue-collar comparators and AI-exposed digital occupations.

**Datasets:** —  
**Curated inputs:** `micro/bls_occupations_physical_vs_digital_2025`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
o = curated("micro", "bls_occupations_physical_vs_digital_2025")
o = o[o.group != "total"]
for c in ("median_wage", "proj_change_pct", "annual_openings", "employment"):
    o[c] = pd.to_numeric(o[c], errors="coerce")
fig, ax = plt.subplots(figsize=(10, 7))
col = {"physical": "#1e8e3e", "digital": "#d93025", "bluecollar": "#f9ab00"}
for gname, d in o.groupby("group"):
    ax.scatter(d.proj_change_pct, d.median_wage / 1000, s=(d.annual_openings.fillna(2000) / 400), alpha=0.6, label=gname, color=col.get(gname))
for _, r in o.iterrows():
    ax.annotate(r.title[:28], (r.proj_change_pct, r.median_wage / 1000), fontsize=6)
ax.axvline(3.5, color="#999", ls="--", lw=0.8); ax.axhline(50.98, color="#999", ls="--", lw=0.8)
ax.set_xlabel("Projected employment change 2025–35, %"); ax.set_ylabel("Median annual wage, $000")
ax.set_title("Wage vs projected growth (bubble = annual openings)", loc="left", fontweight="bold"); ax.legend(frameon=False);
```
