# OEWS: the full occupational wage distribution

Parses the latest national OEWS release (all ~830 detailed occupations) from the raw vintage: employment, median and percentile wages.

**Datasets:** `bls/oews_national`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["bls/oews_national"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
import zipfile
z = zipfile.ZipFile(raw_payload("bls/oews_national"))
print(z.namelist())
```

```python
name = next(n for n in z.namelist() if n.lower().endswith((".xlsx", ".xls")))
oews = pd.read_excel(z.open(name))
oews.columns = [c.lower() for c in oews.columns]
det = oews[oews.o_group == "detailed"].copy()
for c in ("tot_emp", "a_median", "a_pct90"):
    det[c] = pd.to_numeric(det[c], errors="coerce")
det[["occ_code", "occ_title", "tot_emp", "a_median", "a_pct90"]].sort_values("tot_emp", ascending=False).head(25)
```
