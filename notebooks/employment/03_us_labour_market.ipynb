# United States: labour market and sector employment

Headline CPS and CES series since 1990, and the reallocation of payroll employment across manufacturing, construction, information and health care.

**Datasets:** `bls/headline_monthly`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["bls/headline_monthly"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
h = load("bls/headline_monthly")
ur = pivot(select(h, indicator=["LNS14000000", "LNS14000012"]), columns="indicator_label")
plot_lines(ur, "US unemployment rate", "%", "BLS CPS", names={});
```

## Sector employment, index 1995 = 100

```python
sec = ["CES3000000001", "CES2000000001", "CES5000000001", "CES6562000001", "CES0000000001"]
s = pivot(select(h, indicator=sec), columns="indicator_label")
base = s.loc[s.index.str.startswith("1995")].mean()
plot_lines(s / base * 100, "Payroll employment by sector, 1995 = 100", "index", "BLS CES", names={});
```

## Share of nonfarm employment

```python
tot = pivot(select(h, indicator="CES0000000001"), columns="indicator")["CES0000000001"]
share = s.drop(columns=[c for c in s.columns if "Total" in c]).div(tot, axis=0) * 100
share.iloc[[0, -1]].T.rename(columns=lambda c: f"share {c}")
```
