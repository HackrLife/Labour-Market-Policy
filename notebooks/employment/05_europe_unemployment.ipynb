# Europe: unemployment, total and under-25

Eurostat monthly unemployment (seasonally adjusted, % of labour force) for total and under-25 populations.

**Datasets:** `eurostat/une_rt_m`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["eurostat/une_rt_m"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
e = load("eurostat/une_rt_m")
e = select(e, s_adj="SA", sex="T", unit="PC_ACT")
areas = ["EA20", "EU27_2020", "DE", "FR", "IT", "ES", "SE", "PL", "NL"]
latest(select(e, age="Y_LT25", ref_area=areas))
```

```python
plot_lines(pivot(select(e, age="Y_LT25", ref_area=areas)).loc["2005":], "Under-25 unemployment rate (SA)", "%", "Eurostat une_rt_m", names={});
```

```python
plot_lines(pivot(select(e, age="TOTAL", ref_area=areas)).loc["2005":], "Unemployment rate, total (SA)", "%", "Eurostat une_rt_m", names={});
```
