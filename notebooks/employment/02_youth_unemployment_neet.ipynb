# Youth unemployment and NEET

How far young people trail the overall labour market: youth (15–24) unemployment, the youth-to-total ratio, and the share of youth not in employment, education or training.

**Datasets:** `worldbank/unemployment_youth`, `worldbank/unemployment_total`, `ilostat/neet_youth_modelled`  
**Curated inputs:** `employment/youth_labour_markets`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["worldbank/unemployment_youth", "worldbank/unemployment_total", "ilostat/neet_youth_modelled"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
y = pivot(select(load("worldbank/unemployment_youth"), ref_area=FOCUS))
t = pivot(select(load("worldbank/unemployment_total"), ref_area=FOCUS))
plot_lines(y, "Youth unemployment rate (15–24), %", "%", "ILO modelled estimates via World Bank WDI (SL.UEM.1524.ZS)");
```

## Youth-to-total unemployment ratio
A ratio above 2 means a young person is more than twice as likely to be unemployed as the average worker.

```python
ratio = (y / t).dropna(how="all")
plot_bars(ratio.ffill().iloc[-1], f"Youth / total unemployment ratio, {ratio.index[-1]}", "ratio", "World Bank WDI");
```
