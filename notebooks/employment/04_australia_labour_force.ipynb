# Australia: Labour Force

ABS Labour Force headline estimates. The first cells list the dimensions and codes in the latest vintage so the series selection below can be checked against the ABS data structure.

**Datasets:** `abs/labour_force`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["abs/labour_force"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
lf = load("abs/labour_force")
dims = [c for c in lf.columns if c not in ("source", "dataset", "value", "period", "ref_area", "indicator")]
{d: lf[d].drop_duplicates().head(12).tolist() for d in dims}
```

## Unemployment and participation
Set the codes below from the dimension listing (ABS LF: MEASURE M13 = unemployment rate, M12 = participation rate; TSEST 20 = seasonally adjusted; SEX 3 = persons; AGE 1599 = 15+, 1524 = 15–24). Confirm against the listing above if ABS revises its codelists.

```python
SEL = dict(measure="M13", tsest=20, sex=3, region="AUS")
sel = {k: v for k, v in SEL.items() if k in lf.columns}
u = select(lf, **sel)
age_col = "age" if "age" in u.columns else None
p = pivot(u, columns=age_col) if age_col else pivot(u, columns="indicator")
plot_lines(p.tail(240), "Australia: unemployment rate (seasonally adjusted)", "%", "ABS Labour Force, Australia", names={});
```
