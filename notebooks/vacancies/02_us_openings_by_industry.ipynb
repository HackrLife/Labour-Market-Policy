# United States: job openings by industry

Openings rates by industry from JOLTS: where unmet labour demand sits relative to the economy-wide rate.

**Datasets:** `bls/jolts`  
**Curated inputs:** `vacancies/jolts_by_industry_2025_2026`

This notebook reads the latest vintage each time it runs; re-run after a new release to update every table and chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
from labobs.load import load, vintage, curated, raw_payload
from labobs.nb import select, pivot, latest, plot_lines, plot_bars, to_annual, FOCUS, FOCUS_NAMES
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
DATASETS = ["bls/jolts"]
for k in DATASETS:
    v = vintage(k)
    print(f"{k:50} vintage {v.get('vintage', 'not yet ingested'):12} sha {v.get('sha256', '')[:12]}")
```

```python
j = load("bls/jolts")
rates = pivot(select(j, indicator=[c for c in j.indicator.unique() if c.endswith("JOR")]), columns="indicator_label")
plot_lines(rates.loc["2015":], "Job openings rate by industry", "%", "BLS JOLTS", names={});
```

```python
rates.iloc[-1].sort_values(ascending=False).to_frame(rates.index[-1])
```
